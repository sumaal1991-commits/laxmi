**triple quotes**

- in jupyter notebook we have a markdown option to write an information

- **markdown** means we are conveying some information

- but in other platforms like vs code or pycharm we don't have markdown option we can convey using triple quotes

- this is also called as **doc string**

- the code part we can provide single or double quotes

- but conveying the info about the code is **triple quotes**

- **triple quotes** are using in prompt engineering
